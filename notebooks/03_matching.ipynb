```python
import pandas as pd
import numpy as np
```

```python
from sentence_transformers import SentenceTransformer, util
resume_df = pd.read_csv('../data/archive/Resume/Resume.csv')
jobs_df = pd.read_csv("../data/archive/job_title_des.csv")

model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
resume_df.head(5)
```

```python
jobs_df.head(5)
```

```python
resume_text = resume_df['Resume_str'].iloc[0]
job_text = jobs_df['Job Description'].iloc[0]

resume_embedding = model.encode(resume_text)
job_embedding = model.encode(job_text)

similarity_score = util.cos_sim(resume_embedding, job_embedding)

print(f"Resume Category: {resume_df['Category'].iloc[0]}")
print(f"Job Title: {jobs_df['Job Title'].iloc[0]}")
print(f"Similarity Score: {similarity_score.item():.2f}")
```

```python
it_resume = resume_df[resume_df['Category'] == 'INFORMATION-TECHNOLOGY']
resume_text2 = it_resume['Resume_str'].iloc[0]
resume_embedding_it = model.encode(resume_text2)
similarity_score_2 = util.cos_sim(resume_embedding_it, job_embedding)
print(f"Resume Category: IT")
print(f"Job Title: {jobs_df['Job Title'].iloc[0]}")
print(f"Similarity Score: {similarity_score_2.item():.2f}")
```
