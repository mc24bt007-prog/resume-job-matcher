```python
from collections import Counter

# Step 1: count how many skills each word appears in (across the whole dataset)
word_doc_frequency = Counter()

for skill in all_skills:
    words_in_skill = set(skill.split())  # set() so a word is counted once per skill, even if repeated
    for word in words_in_skill:
        word_doc_frequency[word.lower()] += 1

# Step 2: peek at the most common words across all skills
print(word_doc_frequency.most_common(20))
```

```python
# how many words appear in more than N different skills, for a few candidate thresholds
for threshold in [5, 10, 20, 50, 100]:
    count = sum(1 for freq in word_doc_frequency.values() if freq > threshold)
    print(f"Words appearing in more than {threshold} skills: {count}")
```

```python
# 1. Updated function: stopwords + document-frequency filtering
STOPWORDS = {"and", "or", "the", "a", "an", "of", "for", "in", "on", "to", "with", "by", "at"}
DOC_FREQ_THRESHOLD = 20

def get_skill_variants(skill):
    variants = [skill]
    words = skill.split()
    if len(words) > 1:
        meaningful_words = [
            w for w in words
            if w.lower() not in STOPWORDS
            and len(w) > 2
            and word_doc_frequency[w.lower()] <= DOC_FREQ_THRESHOLD
        ]
        variants.extend(meaningful_words)
    return variants

# 2. Fresh matcher
matcher = PhraseMatcher(nlp.vocab, attr="LOWER")

# 3. Regenerate variants
skill_to_variants = {}
all_variant_texts = []

for skill in all_skills:
    variants = get_skill_variants(skill)
    skill_to_variants[skill] = variants
    all_variant_texts.extend(variants)

print("Total variants:", len(all_variant_texts))

# 4. Batch-convert to Docs
variant_docs = list(nlp.pipe(all_variant_texts))
print("Total docs:", len(variant_docs))

# 5. Map Docs back to skills
from collections import defaultdict

skill_to_docs = defaultdict(list)
doc_index = 0

for skill in all_skills:
    num_variants = len(skill_to_variants[skill])
    docs_for_this_skill = variant_docs[doc_index : doc_index + num_variants]
    skill_to_docs[skill] = docs_for_this_skill
    doc_index += num_variants

# 6. Register with matcher
for skill, docs in skill_to_docs.items():
    matcher.add(skill, docs)

print("Skills registered:", len(matcher))

# 7. Re-test on the same resume
doc = nlp(sample_resume)
matches = matcher(doc)
print("Matches:", len(matches))
```

```python
matched_skills = [nlp.vocab.strings[match_id] for match_id, start, end in matches]
skill_counts = Counter(matched_skills)

print(len(skill_counts))
print(skill_counts.most_common(20))
```
