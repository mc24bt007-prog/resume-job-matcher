```python
import pandas as pd
import numpy as np
```

```python
jobs_df = pd.read_csv('../data/archive/job_title_des.csv')
print(jobs_df.shape)
jobs_df.head()
```

```python
import spacy
from spacy.matcher import PhraseMatcher
nlp = spacy.load('en_core_web_sm')
skills_list = [
     # Programming languages
    "Python", "Java", "JavaScript", "TypeScript", "C++", "C#", "SQL", "R", "Swift", "Kotlin", "PHP", "Ruby", "Go", "Dart",
    # Web/Mobile frameworks
    "React", "Angular", "Vue.js", "Node.js", "Flutter", "Django", "Flask", "Spring Boot", "Express.js", "React Native", ".NET",
    # Data/ML
    "Machine Learning", "Deep Learning", "Data Analysis", "Data Science", "NLP", "Computer Vision",
    "TensorFlow", "PyTorch", "Pandas", "NumPy", "Scikit-learn",
    # Cloud/DevOps
    "AWS", "Azure", "Google Cloud", "Docker", "Kubernetes", "Git", "CI/CD", "Jenkins", "Linux",
    # Databases
    "MySQL", "PostgreSQL", "MongoDB", "Firebase", "Redis",
    # Design
    "UI/UX Design", "Figma", "Photoshop", "Adobe XD",
    # Business/Soft skills
    "Project Management", "Agile", "Scrum", "Customer Service", "Team Management", "Communication",
    "Leadership", "Problem Solving", "Marketing", "Sales", "Accounting", "Excel", "Nursing", "Teaching", "REST API"
]

matcher = PhraseMatcher(nlp.vocab, attr='LOWER')
patterns = [nlp.make_doc(skill) for skill in skills_list]
matcher.add("SKILLS", patterns)
```

```python

doc = nlp(jobs_df['Job Description'].iloc[0])
matches = matcher(doc)

found_skills = set()
for match_id, start, end in matches:
    found_skills.add(doc[start:end].text.title())

print(jobs_df['Job Title'].iloc[0])
print(found_skills)
```

```python
for i in range(5):
    doc = nlp(jobs_df['Job Description'].iloc[i])
    matches = matcher(doc)
    found_skills = set(doc[start:end].text.title() for match_id, start, end in matches)
    print(jobs_df['Job Title'].iloc[i], "->", found_skills)
```
